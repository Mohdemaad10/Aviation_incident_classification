```python
import pandas as pd
import sqlite3 
```

```python
#load csv
df=pd.read_csv("clean_data.csv")
pre_df=pd.read_csv("combined_dataset.csv", encoding = "latin1")
```

```python
#create sqlite databse
conn=sqlite3.connect("Aviation.db")
cursor= conn.cursor()

#add id column
df.insert(0, "ID", range(1,len(df)+1))

# move dataset to db
df.to_sql("reports", conn, if_exists= "replace", index= False)

pre_df.to_sql("main_dataset", conn, if_exists= "replace", index = False)
```

```python
#save, verify and close

conn.commit()

cursor.execute("SELECT COUNT(*) from reports")
print("count of reports are", cursor.fetchone()[0])

cursor.execute("SELECT * from reports limit 5")

for row in cursor.fetchall():
    print(row)


conn.close()
```
